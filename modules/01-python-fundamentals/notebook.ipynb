## 2 · Values and data types

A **value** is one piece of information. A **data type** tells Python what kind of information it is and which operations are valid.

| Type | Meaning | Examples |
|---|---|---|
| `int` | whole number | `42`, `-7`, `0` |
| `float` | decimal number | `3.14`, `-0.5`, `2.0` |
| `str` | text in quotes | `"hello"`, `'ML'`, `"42"` |
| `bool` | truth value | `True`, `False` |
| `NoneType` | absence / missing-value marker | `None` |

A value written directly in code is a **literal**.

````python
print(42, type(42))
print(3.14, type(3.14))
print("42", type("42"))
print(True, type(True))
print(None, type(None))
````

## 3 · Variables and assignment

A **variable** is a name that refers to a value. The `=` symbol is the **assignment operator**. Read `age = 25` as “age gets 25,” not as a mathematical equation.

Python first evaluates the right side, then attaches the name on the left to the result.

````python
learner_name = "Maya"
age = 25
course_progress = 0.10
is_beginner = True

print(learner_name)
print(age)
print(course_progress)
print(is_beginner)
````

Good variable names describe meaning. Python style uses **snake_case**: lowercase words separated by underscores.

- Clear: `price_aed`, `discount_rate`, `total_minutes`
- Unclear: `x`, `stuff`, `thing2`

A name may contain letters, digits, and underscores, but cannot begin with a digit, contain spaces, or be a Python keyword such as `if` or `for`. Names are case-sensitive.

````python
score = 10
print("Before:", score)

score = score + 5
print("After:", score)
````

## 5 · Strings and f-strings

A string is text inside matching single or double quotation marks. An **escape sequence** begins with `\` and represents a special character: `\n` is a newline and `\t` is a tab.

````python
print("First line\nSecond line")
print("Name\tScore")
print("Maya\t92")
print("A backslash looks like this: \\")
````

An **f-string** begins with `f`. Python evaluates expressions inside `{curly braces}` and inserts their values into the text.

````python
learner_name = "Maya"
score = 92
maximum = 100

print(f"{learner_name} scored {score} out of {maximum}.")
print(f"Percentage: {score / maximum * 100:.1f}%")
````

The format instruction `:.1f` displays one digit after the decimal point. It changes the display, not the stored value.

### Useful built-in functions and return values

A function's **return value** is the result it gives back. That result can participate in a larger expression or be stored in a variable.

- `len("Python")` returns the number of characters.
- `abs(-7)` returns the absolute distance from zero.
- `round(3.14159, 2)` rounds for a requested number of decimal places.
- `min(...)` and `max(...)` return the smallest and largest arguments.

````python
print(len("Python"))
print(abs(-7))
print(round(3.14159, 2))
print(min(8, 3, 12))
print(max(8, 3, 12))

saved = print("print displays this text")
print("The return value from print is:", saved)
````

### Basic string methods

A **method** is a function associated with a value. Call it with dot notation: `value.method()`. Strings are **immutable**, so these methods return new strings rather than changing the original.

````python
raw_name = "  Maya Ahmed  "
clean_name = raw_name.strip()

print(f"Original: [{raw_name}]")
print(f"Stripped: [{clean_name}]")
print(clean_name.upper())
print(clean_name.lower())
print(clean_name.replace("Maya", "Sara"))
print("Original is still:", raw_name)
````

## 6 · Operators and expressions

An **operator** performs an operation. The values it acts on are **operands**. An **expression** is code that evaluates to a value.

| Operator | Meaning | Example |
|---|---|---|
| `+` | addition | `7 + 2` → `9` |
| `-` | subtraction | `7 - 2` → `5` |
| `*` | multiplication | `7 * 2` → `14` |
| `/` | true division | `7 / 2` → `3.5` |
| `//` | floor division | `7 // 2` → `3` |
| `%` | remainder | `7 % 2` → `1` |
| `**` | exponentiation | `7 ** 2` → `49` |

````python
print("7 + 2  =", 7 + 2)
print("7 - 2  =", 7 - 2)
print("7 * 2  =", 7 * 2)
print("7 / 2  =", 7 / 2)
print("7 // 2 =", 7 // 2)
print("7 % 2  =", 7 % 2)
print("7 ** 2 =", 7 ** 2)
````

### Precedence

Python evaluates parentheses first, then exponentiation, then `*`, `/`, `//`, and `%`, then `+` and `-`. Use parentheses when they make your intention clearer.

````python
# Predict both answers before running.
print(2 + 3 * 4)
print((2 + 3) * 4)
````

For strings, `+` joins and `*` repeats. Python will not silently add text and a number.

````python
full_name = "Ada" + " " + "Lovelace"
divider = "-" * 20
print(full_name)
print(divider)

# This would raise TypeError, so it stays commented:
# print("Age: " + 25)
print(f"Age: {25}")
````

### Augmented assignment

`count += 1` is a shorter form of `count = count + 1`. Similar forms include `-=`, `*=`, and `/=`.

````python
count = 0
count += 1
count += 1
count *= 10
print(count)
````

## 7 · Comparisons and Boolean logic

Comparison operators ask questions and produce `True` or `False`:

- `==` equal to
- `!=` not equal to
- `>` greater than
- `<` less than
- `>=` greater than or equal to
- `<=` less than or equal to

**Remember:** `=` assigns; `==` compares.

````python
score = 75
print(score == 75)
print(score != 75)
print(score >= 50)
print(score < 50)
````

Logical operators combine or reverse conditions:

- `A and B` is true only if both are true.
- `A or B` is true if at least one is true.
- `not A` reverses the truth value.

````python
age = 20
has_id = True

print("Can enter:", age >= 18 and has_id)
print("Needs help:", age < 18 or not has_id)
print("Not an adult:", not (age >= 18))
````

### Truthiness

`False`, `None`, numeric zero, and empty strings are **falsy**. Most other values are **truthy**. A non-empty string is truthy even if its text is `"False"`.

````python
print(bool(0))
print(bool(""))
print(bool(None))
print(bool(12))
print(bool("False"))
````

### Membership and chained comparisons

`in` and `not in` test membership. Python also supports readable chained comparisons such as `18 <= age < 65`. To test specifically for the missing-value marker, use `value is None`.

````python
message = "machine learning"
print("learn" in message)
print("Python" not in message)

age = 25
print(18 <= age < 65)

middle_name = None
print(middle_name is None)
````

## 8 · Decisions with `if`, `elif`, and `else`

A **conditional** selects a branch. The colon starts a block; four spaces indent the statements that belong to it.

````python
temperature = 22

if temperature >= 30:
    label = "hot"
elif temperature >= 15:
    label = "mild"
elif temperature >= 0:
    label = "cold"
else:
    label = "freezing"

print(label)
````

Only the **first true branch** in an `if` / `elif` / `else` chain runs. Change `temperature` to `35`, `8`, and `-4`, running the cell after each change.

````python
# Separate if statements ask separate questions, so both can run.
number = 12

if number > 0:
    print("positive")

if number % 2 == 0:
    print("even")
````

## 9 · Repetition with loops

A **loop** repeats a block. One pass is an **iteration**.

- `while` repeats while a condition stays true.
- `for` repeats once per value from an iterable such as `range()`.

**Choice rule:** use `for` for known values/items or a known number of repetitions. Use `while` when a changing condition decides when to stop, especially when the number of iterations is unknown. If either works, choose the version whose stopping rule is clearest.

````python
count = 3

while count > 0:
    print(count)
    count -= 1

print("Go!")
````

Before running a `while` loop, answer two safety questions:

1. Which value changes each iteration?
2. Why will that change eventually make the condition false?

If nothing makes the condition false, the program has an **infinite loop**. Use the notebook's stop/interrupt button.

````python
print("range(5):")
for n in range(5):
    print(n, end=" ")

print("\n\nrange(2, 6):")
for n in range(2, 6):
    print(n, end=" ")

print("\n\nrange(2, 9, 2):")
for n in range(2, 9, 2):
    print(n, end=" ")

print("\n\nrange(5, 0, -1):")
for n in range(5, 0, -1):
    print(n, end=" ")
````

`range()` includes the start and excludes the stop. Read it as “start here, stop before there.”

A string is also iterable. A `for` loop can receive one character at a time. Module 02 extends the same pattern to lists, dictionaries, and other containers.

````python
for letter in "ML":
    print(letter)
````

### Accumulators and counters

An **accumulator** collects a result across iterations. Initialize it before the loop and update it inside.

````python
total = 0

for number in range(1, 6):
    total += number
    print(f"After adding {number}, total is {total}")

print("Final total:", total)
````

### `break`, `continue`, and nesting

`break` exits the nearest loop. `continue` skips to the next iteration. A block inside another block is **nested** and adds another indentation level.

````python
for number in range(1, 8):
    if number == 3:
        continue
    if number == 6:
        break
    print(number)
````

````python
for number in range(1, 6):
    if number % 2 == 0:
        print(f"{number} is even")
    else:
        print(f"{number} is odd")
````

## 12 · Mini-project: study-session report

The HTML lesson used the same number of minutes each day. This version accepts a different value on each day. It uses only concepts taught in this module.

The interactive `input()` lines are included but commented. The sample data lets the notebook run unattended. To make it interactive, comment out `minutes_today = sample_minutes` and uncomment the `input()` line.

````python
learner_name = "Maya"
planned_days = 5
total_minutes = 0

print(f"Study log for {learner_name}")
print("-" * 24)

for day in range(1, planned_days + 1):
    sample_minutes = 20 + day * 5
    minutes_today = sample_minutes
    # minutes_today = int(input(f"Minutes studied on day {day}: "))

    total_minutes += minutes_today
    print(f"Day {day}: {minutes_today} minutes")

average_minutes = total_minutes / planned_days

if average_minutes >= 45:
    habit = "strong"
elif average_minutes >= 20:
    habit = "building"
else:
    habit = "start small"

print()
print(f"Total: {total_minutes} minutes")
print(f"Hours: {total_minutes / 60:.2f}")
print(f"Daily average: {average_minutes:.1f} minutes")
print(f"Habit: {habit}")
````
